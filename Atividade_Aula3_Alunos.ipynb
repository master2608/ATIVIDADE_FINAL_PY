A tabela deverá ser dividida em 3

- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática

Exemplo 1

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

# Existe algum estado em que a média de mulheres com graduação está abaixo que homens?

```python
# Percentual de mulheres no Brasil; estados abaixo desse valor
pct_brasil = formacao_total.loc[
    formacao_total["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()

regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
estados = formacao_total[~formacao_total["uf_regiao"].isin(regioes)].copy()

abaixo_brasil = estados[estados["percentual_mulheres"] < pct_brasil]

print(f"Percentual de mulheres no Brasil: {pct_brasil}%")
print(f"Estados abaixo do Brasil: {len(abaixo_brasil)}")
abaixo_brasil[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres") 
```

Exemplo 2 — Ciência, Tecnologia, Engenharias e Matemática (STEM)

```python
formacao_stem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
formacao_stem.columns = ["uf_regiao", "total", "homens", "mulheres"]
for col in ["total", "homens", "mulheres"]:
    formacao_stem[col] = pd.to_numeric(formacao_stem[col], errors="coerce")
formacao_stem["percentual_mulheres"] = (
    formacao_stem["mulheres"] / formacao_stem["total"] * 100
).round(2)

pct_brasil_stem = formacao_stem.loc[
    formacao_stem["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()
estados_stem = formacao_stem[~formacao_stem["uf_regiao"].isin(regioes)].copy()
abaixo_stem = estados_stem[estados_stem["percentual_mulheres"] < pct_brasil_stem]

print(f"Percentual de mulheres no Brasil (STEM): {pct_brasil_stem}%")
print(f"Estados abaixo do Brasil: {len(abaixo_stem)}")
abaixo_stem[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres")
```

Exemplo 3 — Educação, Serviços pessoais, Saúde e Bem-estar

```python
formacao_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
formacao_saude.columns = ["uf_regiao", "total", "homens", "mulheres"]
for col in ["total", "homens", "mulheres"]:
    formacao_saude[col] = pd.to_numeric(formacao_saude[col], errors="coerce")
formacao_saude["percentual_mulheres"] = (
    formacao_saude["mulheres"] / formacao_saude["total"] * 100
).round(2)

pct_brasil_saude = formacao_saude.loc[
    formacao_saude["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()
estados_saude = formacao_saude[~formacao_saude["uf_regiao"].isin(regioes)].copy()
abaixo_saude = estados_saude[estados_saude["percentual_mulheres"] < pct_brasil_saude]

print(f"Percentual de mulheres no Brasil (Educação/Saúde): {pct_brasil_saude}%")
print(f"Estados abaixo do Brasil: {len(abaixo_saude)}")
abaixo_saude[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres")
```

## Gráficos com matplotlib e seaborn

Comparamos **graduação no total**, **STEM** e **saúde/educação**: barras, frequência e box plot.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="Set2")

regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
estados_total = formacao_total[~formacao_total["uf_regiao"].isin(regioes)].copy()
estados_stem = formacao_stem[~formacao_stem["uf_regiao"].isin(regioes)].copy()
estados_saude = formacao_saude[~formacao_saude["uf_regiao"].isin(regioes)].copy()
```

```python
# Barras: % de mulheres com graduação (total) por UF
fig, ax = plt.subplots(figsize=(9, 8))
ordem = estados_total.sort_values("percentual_mulheres", ascending=False)
sns.barplot(data=ordem, y="uf_regiao", x="percentual_mulheres", ax=ax, color="steelblue")
ax.axvline(50, color="gray", linestyle="--", label="50%")
ax.set_title("% de mulheres entre pessoas com graduação, por UF")
ax.set_xlabel("% mulheres")
ax.set_ylabel("UF")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Barras agrupadas: homens x mulheres no Brasil (três recortes)
brasil = pd.DataFrame(
    {
        "area": ["Graduação (total)", "STEM", "Saúde / educação"],
        "homens": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_homens"].item(),
            formacao_stem.loc[formacao_stem["uf_regiao"] == "Brasil", "homens"].item(),
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "homens"].item(),
        ],
        "mulheres": [
            formacao_total.loc[formacao_total["uf_regiao"] == "Brasil", "total_mulheres"].item(),
            formacao_stem.loc[formacao_stem["uf_regiao"] == "Brasil", "mulheres"].item(),
            formacao_saude.loc[formacao_saude["uf_regiao"] == "Brasil", "mulheres"].item(),
        ],
    }
)
brasil_long = brasil.melt(id_vars="area", var_name="sexo", value_name="pessoas")

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```

```python
# Histograma (frequência) do % de mulheres nas UFs — total x STEM
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
sns.histplot(estados_total["percentual_mulheres"], bins=8, kde=True, ax=axes[0])
axes[0].axvline(50, color="gray", linestyle="--")
axes[0].set_title("Frequência: % mulheres (graduação total)")
axes[0].set_xlabel("% mulheres")
axes[0].set_ylabel("Número de UFs")

sns.histplot(estados_stem["percentual_mulheres"], bins=8, kde=True, ax=axes[1], color="coral")
axes[1].axvline(50, color="gray", linestyle="--")
axes[1].set_title("Frequência: % mulheres (STEM)")
axes[1].set_xlabel("% mulheres")
plt.tight_layout()
plt.show()
```

```python
# Box plot: comparar a distribuição do % de mulheres nas três áreas
box_df = pd.DataFrame(
    {
        "Graduação (total)": estados_total["percentual_mulheres"].values,
        "STEM": estados_stem["percentual_mulheres"].values,
        "Saúde / educação": estados_saude["percentual_mulheres"].values,
    }
).melt(var_name="area", value_name="percentual_mulheres")

fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=box_df, x="area", y="percentual_mulheres", ax=ax)
ax.axhline(50, color="gray", linestyle="--")
ax.set_title("Box plot: % de mulheres nas UFs, por área de formação")
ax.set_xlabel("Área")
ax.set_ylabel("% mulheres")
plt.tight_layout()
plt.show()
```

```python
# Frequência: UFs com mais mulheres ou mais homens na graduação (total)
estados_total["maioria"] = "Mais mulheres"
estados_total.loc[
    estados_total["total_mulheres"] <= estados_total["total_homens"], "maioria"
] = "Mais homens"

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=estados_total, x="maioria", ax=ax)
ax.set_title("Frequência de UFs segundo a maioria na graduação")
ax.set_xlabel("")
ax.set_ylabel("Número de UFs")
plt.tight_layout()
plt.show()
```
